# Lab Instructions

You are part of a data team for a financial institution that uses an AI model to determine who will and will not be granted a loan.  Recently, there have been complaints from customers that the model unfairly discriminates based on the applicant's age and sex. A court has ordered that the financial institution must provide information about how the model works.  

Data from recent loan applications is given in `loan_data.csv`.  The data dictionary is below:

| Feature              | Description | Data Type |
| :---------------- | :------: | ----: |
|person_age|Age of the person|Float
|person_gender|Gender of the person|Categorical
|person_education|Highest education level|Categorical
|person_income|Annual income|Float
|person_emp_exp|Years of employment experience|Integer
|person_home_ownership|Home ownership status (e.g., rent, own, mortgage)|Categorical
|loan_amnt|Loan amount requested|Float
|loan_intent|Purpose of the loan|Categorical
|loan_int_rate|Loan interest rate|Float
|loan_percent_income|Loan amount as a percentage of annual income|Float
|cb_person_cred_hist_length|Length of credit history in years|Float
|credit_score|Credit score of the person|Integer
|previous_loan_defaults_on_file|Indicator of previous loan defaults|Categorical
|loan_status (target variable)|Loan approval status: 1 = approved; 0 = rejected|Integer

Visualize the relationship between age, sex and 5 other features with `loan_status`.  Which features seem to drive the model's decision to grant a loan?  Which seem unrelated?  Should the financial institution be concerned about potential legal trouble?  Make sure to justify your answers with specific references to your visualizations.

**You must use quantitative visualizations (boxplots or histograms) when comparing quantitative features.  DO NOT categorize quantitative features.**

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("loan_data.csv")

df.head()
```

```python
# Age Approval Rate (Quantitive) - Box Plot

df.boxplot(column="person_age", by="loan_status", vert=False)

plt.title("Age vs Loan Approval")
plt.suptitle("")
plt.xlabel("Loan Status (0 = Rejected, 1 = Approved)")
plt.ylabel("Age")
plt.show()
```

```python
# Gender Approval Rate (Qualitive) - Bar Chart

gender_approval = df[df["loan_status"] == 1].groupby("person_gender")["loan_status"].count()

gender_approval = gender_approval.sort_values(ascending=False)

print(gender_approval)


gender_approval.plot(kind="bar")

plt.title("Gender vs Loan Approval")
plt.xlabel("Gender")
plt.ylabel("Amount Approved")
plt.legend(["Rejected", "Approved"])
plt.show()
```

```python
# Income Rate Feature (Quantitive Feature 1) - Box Plot

df.boxplot(column="person_income", by="loan_status", vert = False)

plt.title("Income vs Loan Approval")
plt.suptitle("")
plt.xlabel("Income")
plt.ylabel("Loan Status")
plt.show()
```

```python
# Credit Score Rate (Quantitive Feature 2) - Historgram

approved = df[df["loan_status"] == 1]["credit_score"]
rejected = df[df["loan_status"] == 0]["credit_score"]

plt.hist(rejected, bins=20, alpha=0.7, label="Rejected")
plt.hist(approved, bins=20, alpha=0.7, label="Approved")

plt.legend()
plt.show()
```

```python
# Education Rate Feature (Qualitive Feature 3) - Bar Chart
education_approval = pd.crosstab(
    df["person_education"],
    df["loan_status"],
    normalize="index"
) * 100

education_approval.plot(
    kind="bar",
    stacked=True
)

plt.title("Loan Status by Education Level")
plt.xlabel("Education Level")
plt.ylabel("Percentage")
plt.legend(["Rejected", "Approved"])

plt.show()
```

```python
# Home Ownership Rate (Qualitive Feature 4) - Bar Chart

home_approval = (
    pd.crosstab(
        df["person_home_ownership"],
        df["loan_status"],
        normalize="index"
    ) * 100
).round(2)

home_approval.plot(kind="bar", stacked = True)

plt.title("Loan Approval Rate by Home Ownership")
plt.xlabel("Home Ownership")
plt.ylabel("Percentage")
plt.legend(["Rejected", "Approved"])

plt.show()
```

```python
# Loan Amount Rate (Quantitive Feature 5) - Box Plot

df.boxplot(column="loan_amnt", by="loan_status", vert=False)

plt.title("Loan Amount vs Loan Approval")
plt.suptitle("")
plt.xlabel("Loan Amount")
plt.ylabel("Loan Status (0 = Rejected, 1 = Approved)")

plt.show()
```

```python
# Person Income Rate (Quantitive Bonus Feature 6) - Box Plot

df.boxplot(column="person_income", by="loan_status", vert=False)

plt.title("Income vs Loan Approval")
plt.suptitle("")
plt.xlabel("Income")
plt.ylabel("Loan Status (0 = Rejected, 1 = Approved)")

plt.show()
```

```python
# Check Difference in Credit Score between Genders (Bonus Qualitive Plot) - Bar Chart
gender_and_credit_score = (df.groupby("person_gender")["credit_score"]
    .mean()
    .round(2)
)

gender_and_credit_score.plot(kind = 'bar')
```

```python
# Check avg Credit Score Between Ages (Bonus Quantitive Plot) Line Plot
age_and_credit_score = (
    df.groupby("person_age")["credit_score"]
    .mean()
    .round(2)
)

age_and_credit_score.plot(kind="line")

plt.title("Average Credit Score by Age")
plt.xlabel("Age")
plt.ylabel("Average Credit Score")

plt.show()
```

```python
# Check Difference in Ownership between Genders (Bonus Qualitive Plot) - Bar Chart
gender_home = pd.crosstab(
    df["person_home_ownership"],
    df["person_gender"]
)

gender_home.plot(kind="bar")

plt.title("Gender Distribution by Home Ownership")
plt.xlabel("Home Ownership")
plt.ylabel("Count")

plt.show()
```

```python
# Check difference in Age and Home Ownership (Bonus Quantitive Plot) - Box Plot
df.boxplot(column="person_age", by="person_home_ownership", vert=False)

plt.title("Age Distribution by Home Ownership")
plt.suptitle("")
plt.xlabel("Age")
plt.ylabel("Home Ownership")

plt.show()
```

```python
# Check Difference in Income Between Genders (Bonus Quantitive Plot) - Box Plot

df.boxplot(column="person_income", by="person_gender", vert=False)

plt.title("Income Distribution by Gender")
plt.suptitle("")
plt.xlabel("Income")
plt.ylabel("Gender")

plt.show()
```

```python
# Check difference in Age and Income (Bonus Quantitive Plot) - Scatter Plot
plt.scatter(df["person_age"], df["person_income"])

plt.title("Income vs Age")
plt.xlabel("Age")
plt.ylabel("Income")

plt.show()
```
