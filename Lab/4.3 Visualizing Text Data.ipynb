```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Starbucks_reviews.csv")
```

```python
def setup_graph(title, xlabel, ylabel, fig_size = (8, 5)):
    plt.figure(figsize= fig_size)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)      
```

```python
# Visualization 1 (Qualitive)
rating_counts = df["Rating"].value_counts().sort_index()

setup_graph("Distribution of Starbucks Review Ratings", "Rating (Stars)", "Number of Reviews")
plt.bar(rating_counts.index, rating_counts.values)
plt.xticks([1, 2, 3, 4, 5])
plt.show()
```

The data above displays the rating by stars in accordance with reviews. It informs us that there are many low ratings in the starbucks industry, at least according to this data. This could be bad for the company if they don't analyze their data and see what their customers are saying about the service, products, and performance of the company.

```python
# Visualization 1 (Quantitive)
df["Review_Length"] = df["Review"].astype(str).apply(
    lambda x: len(x.split())
)

setup_graph("Distribution of Review Lengths", "Number of Words", "Number of Reviews")
plt.hist(df["Review_Length"], bins=25)
plt.show()
```

The data above showcases the length of reviews and where they lie. According to this data most reviews lie between the range of 40 to 150 words, which is probably the opposite of what the company wants as compliments and words of acknowledgement are usually given as "Short and Sweet" not long and thorough. This could indicate complaints and/or explanations about Starbucks and their products or service.

```python
# Visualization 3 (Qualitive)
avg_length = df.groupby("Rating")["Review_Length"].mean()

setup_graph("Average Review Length by Rating", "Rating", "Average Number of Words")
avg_length.plot(kind="bar")
plt.xticks(rotation=0)
plt.show()
```
