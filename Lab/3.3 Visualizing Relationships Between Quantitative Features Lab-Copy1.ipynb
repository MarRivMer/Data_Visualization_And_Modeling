```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("Employee_productivity.csv")

df.head()
```

```python
def evaluate_feature(feature_name):
    X = df[[feature_name]]
    y = df["ProductivityScore"]

    # Linear Regression and R^2 Score
    model = LinearRegression()
    model.fit(X, y)

    predictions = model.predict(X)

    r2 = r2_score(y, predictions)

    # Graphing
    plt.figure(figsize=(6,4))
    plt.scatter(X, y)
    plt.xlabel(feature_name)
    plt.ylabel("ProductivityScore")
    plt.title(f"{feature_name} vs productivity")
    plt.show()

    print(f"R^2 for {feature_name}: {r2}")


def plot_categorical(feature_name):

    # Categorial
    averages = df.groupby(feature_name)["ProductivityScore"].mean()

    # Graphing
    plt.figure(figsize=(6,4))

    averages.plot(kind="bar")

    plt.xlabel(feature_name)
    plt.ylabel("Average Productivity")
    plt.title(f"{feature_name} vs Productivity")

    plt.xticks(rotation=45)

    plt.show()
```

```python
# Main Features
evaluate_feature("YearsAtCompany")
evaluate_feature("HoursWorked")
evaluate_feature("FocusScore")
```

```python
# Other Features

evaluate_feature("TeamMeetingsPerWeek")
plot_categorical("Department")
plot_categorical("DeskLocation")
```

The biggest factor by far was focus, showing the highest results between the two, the higher the focus on an individual the more productive they were. The years did not matter which was suprising, and the hours mattered slightly with the sweet range, ranging from 30 to 40 hours.
