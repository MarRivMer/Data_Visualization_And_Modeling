```python
# Setup Data Set
import pandas as pd

df = pd.read_csv("winners_f1_1950_2025_v2.csv")

df.head()
```

```python
# Answer 1: 
max_wins = df[df["winner_name"] == "Max Verstappen"]

max_wins[["grand_prix"]]
```

From the output we can analyze that Max verstappen has won 60+ Grand Prix, in many places such as Spain Malaysia, Mexico, and Austria. Not just one Grand Prix but many.

```python
# Answer 2:
best_lap = max_wins["time"].min()

win = max_wins.loc[max_wins["time"] == best_lap]

win.head()
```

The output tells us that Max Verstappen's best lap was 1:13:41 and it was in the "Autodromo Nazionale di Monza" circuit in Italy

```python
# Answer 3:
max_wins.groupby("year").size().sort_values(ascending=True)
```

```python
He won races every year from 2016 to 2025 an has multiple wins every year besided 2016. The answer to this question is extremely suprising because I did not know this an just goes to show why he is widely considered one of the
greatest racers of all time.
```

```python
# Answer 4:
max_wins.groupby("circuit").size().sort_values(ascending=True)
```

The data analysis tells us that "Autodromo Hermanos Rodriguez" and "Red Bull Ring" are his best tracks in which he has won 5 Grand Prix in both of those tracks.

```python
# Answer 5:
max_wins.groupby("team").size().sort_values(ascending=True)
```

Max Verstappen has won the most with Red "Bull Racing Honda RBPT" team, with 28 total Grand Prix wins!
