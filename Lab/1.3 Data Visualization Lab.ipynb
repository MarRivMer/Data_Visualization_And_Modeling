```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('spotify-2023.csv', encoding='ISO-8859-1')

df.head()
```

```python
df['bpm'].plot(
    kind='hist',
    bins=20,
    title='BPM Distribution',
    xlabel='Beats Per Minute',
    ylabel='Number of Songs'
)

plt.show()
```

```python
df['key'].value_counts().plot(
    kind='bar',
    title='Most Common Song Keys',
    xlabel='Key',
    ylabel='Number of Songs'
)

plt.show()
```

```python
df['mode'].value_counts().plot(
    kind='bar',
    title='Mode Distribution',
    xlabel='Mode',
    ylabel='Number of Songs'
)

plt.show()
```

```python
df['danceability_%'].plot(
    kind='hist',
    bins=20,
    title='Danceability Distribution',
    xlabel='Danceability',
    ylabel='Number of Songs'
)

plt.show()
```

```python
df['energy_%'].plot(
    kind='hist',
    bins=20,
    title='Energy Distribution',
    xlabel='Energy',
    ylabel='Number of Songs'
)

plt.show()
```

```python
df['speechiness_%'].plot(
    kind='hist',
    bins=20,
    title='Speechiness Distribution',
    xlabel='Speechiness',
    ylabel='Number of Songs'
)

plt.show()
```

```python
df['acousticness_%'].plot(
    kind='hist',
    bins=20,
    title='Acousticness Distribution',
    xlabel='Acousticness',
    ylabel='Number of Songs'
)

plt.show()
```

```python
popular_subset = df.loc[
    (df['key'] == 'C#') &
    (df['bpm'] >= 110) &
    (df['bpm'] <= 130) &
    (df['mode'] == 'Major')
]

popular_subset.head()
```
