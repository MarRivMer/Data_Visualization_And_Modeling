```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("titanic_passengers.csv")
```

```python
df.head()
```

```python
# Added Age Group Column to the Data Frame
df["Age Group"] = "Unknown"

df.loc[df["Age"] < 13, "Age Group"] = "Child"
df.loc[(df["Age"] >= 13) & (df["Age"] < 20), "Age Group"] = "Teen"
df.loc[(df["Age"] >= 20) & (df["Age"] < 60), "Age Group"] = "Adult"
df.loc[df["Age"] >= 60, "Age Group"] = "Senior"
```

```python
# Added Fare Group Column to the Data Frame
df["Fare Group"] = "Unknown"

df.loc[df["Fare"] < 20, "Fare Group"] = "Low Fare"
df.loc[(df["Fare"] >= 20) & (df["Fare"] < 50), "Fare Group"] = "Medium Fare"
df.loc[df["Fare"] >= 50, "Fare Group"] = "High Fare"
```

```python
df.head()
```

```python
# Check Survival Chance by Gender
sex_survival = (df.groupby("Sex")["Survived"].mean() * 100).round(2)
sex_survival = sex_survival.sort_values(ascending = False)

print(sex_survival)

sex_survival.plot(kind="pie")

plt.title("Survival Rate by Gender")
plt.show()
```

```python
# Check Survival by Class
pclass_survival = (df.groupby("Pclass")["Survived"].mean() * 100).round(2)
pclass_survival = pclass_survival.sort_values(ascending = False)

print(pclass_survival)


pclass_survival.plot(kind="bar")

plt.title("Survival Rate by Passenger Class")
plt.ylabel("Survival Rate")
plt.show()
```

```python
# Check Survival by Age
age_survival = (df.groupby("AgeGroup")["Survived"].mean() * 100).round(2)
age_survival = age_survival.sort_values(ascending = False)

print(age_survival)


age_survival.plot(kind="bar")

plt.title("Survival Rate by Age Group")
plt.ylabel("Survival Rate")
plt.show()
```

```python
# Check Survival by Fare
fare_survival = (df.groupby("FareGroup")["Survived"].mean() * 100).round(2)
fare_survival = fare_survival.sort_values(ascending = False)

print(fare_survival)


fare_survival.plot(kind="bar")

plt.title("Survival Rate by Fare Group")
plt.ylabel("Survival Rate")
plt.show()
```

```python
# Check Survival  by Embarked Location
embarked_survival = (df.groupby("Embarked")["Survived"].mean() * 100).round(2)
embarked_survival = embarked_survival.sort_values(ascending = False)

print(embarked_survival)


embarked_survival.plot(kind="bar")

plt.title("Survival Rate by Embarked Port")
plt.ylabel("Survival Rate")
plt.show()
```
