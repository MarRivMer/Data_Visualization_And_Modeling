```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("USA_cars_datasets.csv")
```

```python
def setup_graph(title, xlabel, ylabel, fig_size = (8, 5)):
    plt.figure(figsize= fig_size)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
```

```python
# Remove Bad Values
df["price"] = pd.to_numeric(df["price"], errors="coerce")
df["mileage"] = pd.to_numeric(df["mileage"], errors="coerce")
df["year"] = pd.to_numeric(df["year"], errors="coerce")
```

```python
# Visualization 1 (Quantitive)

setup_graph("Car Price vs Mileage", "Mileage", "Price")

plt.scatter(df["mileage"], df["price"], alpha=0.3)
plt.show()
```

Data Analysis - The data above demonstrates the effect mileage has on a cars price. The higher the mileage on a car the lower the price and the oppisite is true. cars with 0 milieage can go up to $80,000 US dollars; however, any car with some mileage even if very minimal drastically decreases in price. Looking into cars with some mileage might be beneficial.

```python
# Visualization 2 (Qualitive)

avg_price_brand = df.groupby("brand")["price"].mean().sort_values(ascending = False).head(10)

setup_graph("Top 10 Brands by Average Price", "Brand", "Average Price", (10,5))

avg_price_brand.plot(kind="bar")
plt.xticks(rotation=45)
plt.show()
```

Data Analysis - The data above shows the average price for the top 10 brands. To my suprise chevorolet are quite affordable if you find the correct model. The data shows us that the brand also can effec the price of a car with Harley-Davidson being the most expensive on average. This; however, does not mean you can't find a Harley-Davidson for an affordable price.

```python
# Visualization 3 (Quantitive)

setup_graph("Car Year vs Price", "Year", "Price in US Dollars")

plt.scatter(df["year"], df["price"], alpha=0.3)
plt.show()
```

Data Analysis - The year of the car effects the target feature quite a bit as well. The newer the model of the car you are looking to buy, the higher the price will be as well. If you are looking for something affordable, looking into older models might be the way to go.

```python
# Visualization 4 (Quantitive)
setup_graph("Distribution of Car Prices", "Price", "Frequency", (10, 5))

plt.hist(df["price"].dropna(), bins=30)
plt.show()
```

Data Analysis - The data above showcases the distribution of car prices with most cars ranging from around $10,000 to $20,000, if you are looking to buy a car make sure you are able to afford within this range or more. 

```python
# Visualization 4 (Qualitive)

avg_title = df.groupby("title_status")["price"].mean()

setup_graph("Average Price by Title Status", "Title Status", "Average Price")

avg_title.plot(kind="bar")
plt.xticks(rotation=45)
plt.show()
```

Data Analysis - The title status shows us how much a vehicle can cost based on its condition. If it is left in pieces then the price will range around $2500 pressuming the car does not work. A clean and working vehicle average cost will be around the $20,000 mark.
